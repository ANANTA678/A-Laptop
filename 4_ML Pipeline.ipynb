```python
# Page URL --> https://www.kaggle.com/datasets/arshmankhalid/bank-marketing-ml-ready-dataset
# Name Data Set --> Bank Marketing ML Ready Dataset

import kagglehub
import pandas as pd
from pathlib import Path

# Download the Kaggle dataset
path = kagglehub.dataset_download(
    "arshmankhalid/bank-marketing-ml-ready-dataset"
)

# Find downloaded CSV and Excel files
dataset_path = Path(path)

files = [
    file for file in dataset_path.rglob("*")
    if file.suffix.lower() in [".csv", ".xlsx", ".xls"]
]

# Display available files
print("Available files:")
for file in files:
    print(file.name)

# Load the first available file
if not files:
    raise FileNotFoundError("No CSV or Excel file found.")

file_path = files[0]

if file_path.suffix.lower() == ".csv":
    df = pd.read_csv(file_path)
else:
    df = pd.read_excel(file_path)

# Display dataset information
print("\nDataset loaded successfully!")
print("File:", file_path.name)
print("Shape:", df.shape)

display(df.head())
display(df.info())
```

### 1. Create the Pipeline base on Data Category

```python
Numerical_data = df.select_dtypes(include='number').columns.tolist()
Ordinal_data = ['Education']
Nominal_Data = df.select_dtypes(include='object').drop(columns=['Education', 'TARGET'], errors='ignore').columns.tolist()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
```

```python
Numerical_pipline = Pipeline(steps=[
    ('Imputer', SimpleImputer(strategy='mean')),
    ('Scaling', StandardScaler())
])
```

```python
Ordinal_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoding', OrdinalEncoder(categories=[['unknown', 'primary', 'secondary', 'tertiary']])),
    ('Scaler', StandardScaler())
])
```

```python
Nominal_Pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoding', OneHotEncoder(handle_unknown='ignore'))
])
```

### 2. Use Column Tranformaction

```python
from sklearn.compose import ColumnTransformer

Features = ColumnTransformer(transformers=[
    ('Numerical Transformation', Numerical_pipline, Numerical_data),
    ('Ordinal Transformation', Ordinal_pipeline, Ordinal_data),
    ('Nominal Transformation', Nominal_Pipeline, Nominal_Data)
])
Features
```

### 3. Create Model Pipeline

```python
from sklearn.linear_model import LogisticRegression

model_pipe = Pipeline( steps=[
    ('ColumnTransformer',Features ),
    ('Classider', LogisticRegression())
])

model_pipe
```

### 4. Split data in Train Test Split

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

Y = df["TARGET"]
le = LabelEncoder()
le.fit_transform(Y)

X = df.drop(columns='TARGET')

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2)
```

```python
model_pipe.fit(X_train, Y_train)
```

```python
Y_predict = model_pipe.predict(X_test)
Y_predict
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y_test, Y_predict)
```

## Model - 2 RandomForestClassifier

```python
from sklearn.ensemble import RandomForestClassifier

pipe_2_model = Pipeline(steps=[
    ('ColumnTransformer', Features ),
    ('Model_classifier', RandomForestClassifier())
])

```

```python
pipe_2_model.fit(X_train, Y_train)

Y_pred_2 = pipe_2_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(Y_test, Y_pred_2)
```
