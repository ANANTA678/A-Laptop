##### 12. Data Formatting

````python
import pandas as pd

df = pd.DataFrame({
    "Name": [" ananta ", "RAHUL", "priya", "Amit "],
    "Date": ["2026/01/15", "2026/02/20", "2026/03/10", "2026/04/25"],
    "Salary": [35000.567, 42000.789, 38000.456, 50000.123],
    "Amount": [1250.5678, 2300.4567, 1800.9876, 4500.1234],
    "Gender": ["M", "F", "Female", "Male"],
    "City": [" pune ", "PUNE", "Mumbai", "mumbai"],
    "Product@": ["Laptop!", "Mouse#", "Keyboard$", "Monitor%"]
})

df
````

````python
## 12. Data Formatting

# 1. **Standardize Date Format**
df["Date"] = pd.to_datetime(df["Date"]).dt.strftime("%d-%m-%Y")
   
# 2. **Standardize Text Case**
df["Name"] = df["Name"].str.title()

# 3. **Standardize Number Format**
df["Salary"] = df["Salary"].round(1)

# 4. **Standardize Decimal Places**
df["Amount"] = df["Amount"].round(2)

## remove Deciman values
df["Amount"] = df["Amount"].astype('int32')

# 5. **Standardize Column Names**
# df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# 6. **Standardize Categorical Values**
df["Gender"] = df["Gender"].replace({"M": "Male","F": "Female"})

df
````
