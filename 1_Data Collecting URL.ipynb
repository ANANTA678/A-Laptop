```python

import pandas as pd
import requests
import time

# Create an empty DataFrame before the loop
final_data = pd.DataFrame()

YOUR_API_KEY = "52bc6c90c263cde3f8518a1c880c803c"

for i in range(1, 501):

    url = f"https://api.themoviedb.org/3/trending/movie/day?api_key={YOUR_API_KEY}&language=en-US&page={i}"

    headers = {
        "accept": "application/json",
        "Authorization": "Bearer YOUR_ACCESS_TOKEN"
    }

    response = requests.get(url, headers=headers, timeout=30)

    if response.status_code == 200:
        json_data = response.json()

        # Convert JSON results into a DataFrame
        df = pd.DataFrame(json_data["results"])

        # Append the current page's data
        final_data = pd.concat(
            [final_data, df],
            ignore_index=True
        )

        print(f"Page {i} completed | Total rows: {len(final_data)}")

    else:
        print(f"Page {i} failed: {response.status_code}")
        print(response.text)

    time.sleep(0.1)

# Display the final dataset
print(final_data.shape)
print(final_data.head())

# Save the data to CSV
final_data.to_csv("trending_movies.csv", index=False)
```
